# Checking first model results (test)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Model data

```python
model_results = pd.read_csv('../results/model_data_test_batch_run.csv')
model_results = model_results.drop(columns = ['Unnamed: 0'])
model_results.head()
```

```python
model_results.weight_opinion_others.unique()
```

```python
model_results.threshold_similarity_bias.unique()
```

```python
model_results.fraction_ties_added_at_flood.unique()
```

## Plotting over time, changing weight_opinion_others
threshold_similarity_bias constant 0.3
fractions_ties_added constant 0.01

```python
weight_opinion_others_model_results = model_results.loc[(model_results.threshold_similarity_bias == 0.3) & (model_results.fraction_ties_added_at_flood == 0.1)]
weight_opinion_others_model_results.head()
```

```python
sns.lineplot(data=weight_opinion_others_model_results, x='Step', y='cumulative_damage', hue='weight_opinion_others')

plt.xlabel("Steps")
plt.ylabel("cumulative_damage")
plt.show()
```

```python
sns.lineplot(data=weight_opinion_others_model_results, x='Step', y='fraction_adapted_any', hue='weight_opinion_others')

plt.xlabel("Steps")
plt.ylabel("fraction_adapted_any")
plt.show()
```

```python
sns.lineplot(data=weight_opinion_others_model_results, x='Step', y='fraction_adapted_both', hue='weight_opinion_others')

plt.xlabel("Steps")
plt.ylabel("fraction_adapted_any")
plt.show()
```

```python
sns.scatterplot(data=weight_opinion_others_model_results, x='Step', y='edges', hue='weight_opinion_others')

plt.xlabel("Steps")
plt.ylabel("nr_edges")
plt.show()
```

## Agent data

```python
agent_results = pd.read_csv('../results/agent_data_test_batch_run.csv')
agent_results = agent_results.drop(columns = ['Unnamed: 0'])
agent_results.head()
```

```python
agent_results.columns
```

### Following one agent
Agent 10

```python
agent_results_agent_10 = agent_results.loc[(agent_results.AgentID == 10) & 
(agent_results.fraction_ties_added_at_flood == 0.1) & (agent_results.threshold_similarity_bias == 0.1) & (agent_results.weight_opinion_others == 0.4) &
(agent_results.seed == 12345)] 
```

```python
sns.lineplot(data=agent_results_agent_10, x='Step', y='overall_intention_DP')
sns.lineplot(data=agent_results_agent_10, x='Step', y='PMT_intention_DP')
sns.lineplot(data=agent_results_agent_10, x='Step', y='SN_intention')

plt.xlabel("Steps")
plt.ylabel("Intention_DP")
plt.show()
```

```python
sns.lineplot(data=agent_results_agent_10, x='Step', y='overall_intention_WP')
sns.lineplot(data=agent_results_agent_10, x='Step', y='PMT_intention_WP')
sns.lineplot(data=agent_results_agent_10, x='Step', y='SN_intention')

plt.xlabel("Steps")
plt.ylabel("Intention_WP")
plt.show()
```

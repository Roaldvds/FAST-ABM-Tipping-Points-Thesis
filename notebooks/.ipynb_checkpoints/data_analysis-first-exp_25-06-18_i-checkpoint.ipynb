# Checking first model results
- 50 replications, 1000 HH, experimental NW, SN all, similarity-biased threat and coping exchange, no NW evolution
- only thing changing (apart from seed): aging of measures (!5, 10, 5, 30)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

## Model data

```python
model_results = pd.read_csv('../results/model_data_250618_i.csv', header = [0, 1, 2])
#model_results = model_results.drop(columns = ['Unnamed: 0'])
model_results.head()
```

```python
model_results.columns = ['_'.join(filter(lambda x: not str(x).startswith('Unnamed:') and not pd.isna(x), map(str, col))) 
              for col in model_results.columns.values]
```

```python
model_results.head()
```

```python
print(model_results.columns.tolist())
```

## Plotting over time

```python
sns.lineplot(data=model_results, x='Step', y='fraction_adapted_any_mean', hue='measures_aging')

plt.xlabel("Steps")
plt.ylabel("Fraction_adapted mean")
plt.grid(True)
plt.show()
```

```python
sns.lineplot(data=model_results, x='Step', y='cumulative_damage_mean', hue='measures_aging')

plt.xlabel("Steps")
plt.ylabel("Damage mean")
plt.grid(True)
plt.show()
```

## Agent data

```python
agent_results = pd.read_csv('../results/agent_data_250618_i.csv', header = [0, 1, 2])
#agent_results = agent_results.drop(columns = ['Unnamed: 0'])
agent_results.columns = ['_'.join(filter(lambda x: not str(x).startswith('Unnamed:') and not pd.isna(x), map(str, col))) 
              for col in agent_results.columns.values]
agent_results.head()
```

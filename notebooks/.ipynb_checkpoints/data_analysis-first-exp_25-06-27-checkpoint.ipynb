# Checking first model results
- 50 replications, 1000 HH, experimental NW, SN all, similarity-biased threat and coping exchange, no NW evolution
- only thing changing (apart from seed): minimum intention needed to consider adapting


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

## Model data

```python
model_results = pd.read_csv('../results/model_data_250623_refined.csv', header = [0, 1, 2])
#model_results = model_results.drop(columns = ['Unnamed: 0'])
model_results.columns = ['_'.join(filter(lambda x: not str(x).startswith('Unnamed:') and not pd.isna(x), map(str, col))) 
              for col in model_results.columns.values]
model_results.head()
```

```python
print(model_results.columns.tolist())
```

```python
columns_to_drop = ['seed_std', 'seed_min', 'seed_max', 'seed_median', 'seed_q1', 'seed_q3', 
                   'nr_households_std', 'nr_households_min', 'nr_households_max', 'nr_households_median', 'nr_households_q1', 'nr_households_q3', 
                   'proportion_ties_added_per_step_std', 'proportion_ties_added_per_step_min', 'proportion_ties_added_per_step_max', 'proportion_ties_added_per_step_median', 'proportion_ties_added_per_step_q1', 'proportion_ties_added_per_step_q3', 
                   'proportion_ties_removed_per_step_std', 'proportion_ties_removed_per_step_min', 'proportion_ties_removed_per_step_max', 'proportion_ties_removed_per_step_median', 'proportion_ties_removed_per_step_q1', 'proportion_ties_removed_per_step_q3' 
                    ]
model_results = model_results.drop(columns = columns_to_drop)
print(model_results.columns.tolist())
```

## Plotting over time

```python
plt.figure(figsize=(10, 6))

# Get unique risk levels for coloring
intentions = model_results['minimum_intention_to_consider_measure'].unique()
colors = plt.cm.viridis(np.linspace(0, 1, len(intentions)))  # Using a colormap

for intention, color in zip(intentions, colors):
    # Filter data for current risk level
    subset = model_results[model_results['minimum_intention_to_consider_measure'] == intention]
    
    # Plot the mean line
    plt.plot(subset['Step'], subset['fraction_adapted_any_mean'], 
             label=f'Intention to adapt {intention}', color=color, linewidth=2)
    
    # Calculate confidence intervals (using 95% CI: mean ± 1.96*std)
    ci_low = subset['fraction_adapted_any_mean'] - 1.96 * subset['fraction_adapted_any_std']
    ci_high = subset['fraction_adapted_any_mean'] + 1.96 * subset['fraction_adapted_any_std']
    
    # Plot the confidence interval as a shaded area
    plt.fill_between(subset['Step'], ci_low, ci_high, 
                    color=color, alpha=0.1)
    
    # Optional: plot quartiles (Q1 and Q3) as dashed lines
    #plt.plot(subset['Step'], subset['money_q1'], 
     #        linestyle='--', color=color, alpha=0.5)
    #plt.plot(subset['Step'], subset['money_q3'], 
     #        linestyle='--', color=color, alpha=0.5)

plt.xlabel('Step', fontsize=12)
plt.ylabel('Fraction adapted', fontsize=12)
#plt.title('F', fontsize=14)
plt.legend(title='Minimum intention to consider measure', fontsize=10, bbox_to_anchor = (1.05, 1), loc = 'upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../reports/figures/intention_to_consider_measure.png')
plt.show()
```

```python
sns.lineplot(data=model_results, x='Step', y='fraction_adapted_any_mean', hue='minimum_intention_to_consider_measure')

plt.xlabel("Steps")
plt.ylabel("Fraction_adapted mean")
plt.grid(True)
plt.show()
```

```python
sns.lineplot(data=model_results, x='Step', y='cumulative_damage_mean', hue='minimum_intention_to_consider_measure')

plt.xlabel("Steps")
plt.ylabel("Damage mean")
plt.grid(True)
plt.show()
```

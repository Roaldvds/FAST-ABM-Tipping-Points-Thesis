# Checking first model results
- 50 replications, 1000 HH, experimental NW, SN all, similarity-biased threat and coping exchange, no NW evolution
- only thing changing (apart from seed): minimum intention needed to consider adapting


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

## Model data

```python
model_results = pd.read_csv('../results/model_data_250623.csv', header = [0, 1, 2])
#model_results = model_results.drop(columns = ['Unnamed: 0'])
model_results.columns = ['_'.join(filter(lambda x: not str(x).startswith('Unnamed:') and not pd.isna(x), map(str, col))) 
              for col in model_results.columns.values]
model_results.head()
```

```python
print(model_results.columns.tolist())
```

```python
columns_to_drop = ['seed_std', 'seed_min', 'seed_max', 'seed_median', 'seed_q1', 'seed_q3', 
                   'nr_households_std', 'nr_households_min', 'nr_households_max', 'nr_households_median', 'nr_households_q1', 'nr_households_q3', 
                   'per_HH_adapted_to_SN_prob_midpoint_std', 'per_HH_adapted_to_SN_prob_midpoint_min', 'per_HH_adapted_to_SN_prob_midpoint_max', 'per_HH_adapted_to_SN_prob_midpoint_median', 'per_HH_adapted_to_SN_prob_midpoint_q1', 'per_HH_adapted_to_SN_prob_midpoint_q3', 
                   'weight_SN_vs_PMT_std', 'weight_SN_vs_PMT_min', 'weight_SN_vs_PMT_max', 'weight_SN_vs_PMT_median', 'weight_SN_vs_PMT_q1', 'weight_SN_vs_PMT_q3', 
                   'weight_opinion_others_std', 'weight_opinion_others_min', 'weight_opinion_others_max', 'weight_opinion_others_median', 'weight_opinion_others_q1', 'weight_opinion_others_q3', 
                    'threshold_similarity_bias_std', 'threshold_similarity_bias_min', 'threshold_similarity_bias_max', 'threshold_similarity_bias_median', 'threshold_similarity_bias_q1', 'threshold_similarity_bias_q3', 
                   'proportion_ties_added_per_step_std', 'proportion_ties_added_per_step_min', 'proportion_ties_added_per_step_max', 'proportion_ties_added_per_step_median', 'proportion_ties_added_per_step_q1', 'proportion_ties_added_per_step_q3', 
                   'proportion_ties_removed_per_step_std', 'proportion_ties_removed_per_step_min', 'proportion_ties_removed_per_step_max', 'proportion_ties_removed_per_step_median', 'proportion_ties_removed_per_step_q1', 'proportion_ties_removed_per_step_q3', 
                    'fraction_ties_added_at_flood_std', 'fraction_ties_added_at_flood_min', 'fraction_ties_added_at_flood_max', 'fraction_ties_added_at_flood_median', 'fraction_ties_added_at_flood_q1', 'fraction_ties_added_at_flood_q3', 
                   'worry_increase_with_flood_std', 'worry_increase_with_flood_min', 'worry_increase_with_flood_max', 'worry_increase_with_flood_median', 'worry_increase_with_flood_q1', 'worry_increase_with_flood_q3',
                   'min_worry_after_flood_std', 'min_worry_after_flood_min', 'min_worry_after_flood_max', 'min_worry_after_flood_median', 'min_worry_after_flood_q1', 'min_worry_after_flood_q3', 
                   'flood_remembrance_period_std', 'flood_remembrance_period_min', 'flood_remembrance_period_max', 'flood_remembrance_period_median', 'flood_remembrance_period_q1', 'flood_remembrance_period_q3']
model_results = model_results.drop(columns = columns_to_drop)
print(model_results.columns.tolist())
```

## Plotting over time

```python
sns.lineplot(data=model_results, x='Step', y='fraction_adapted_any_mean', hue='minimum_intention_to_consider_measure')

plt.xlabel("Steps")
plt.ylabel("Fraction_adapted mean")
plt.grid(True)
plt.show()
```

```python
sns.lineplot(data=model_results, x='Step', y='cumulative_damage_mean', hue='minimum_intention_to_consider_measure')

plt.xlabel("Steps")
plt.ylabel("Damage mean")
plt.grid(True)
plt.show()
```
